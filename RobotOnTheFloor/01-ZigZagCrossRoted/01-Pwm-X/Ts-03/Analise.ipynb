```python
import pandas as pd
import numpy as np

TARGETS = ["DeltaTheta", "DeltaX", "DeltaY"]

SETS = [
    "Train_1",
    "Train_2",
    "Val",
    "Test_1",
    "Test_2",
    "Test_3"
]

results = pd.read_excel("resultados.xlsx")
```

```python
best_models_tables = {}
summary_rows = []  # ← aqui vamos salvar média e desvio

for target in TARGETS:
    for dataset in SETS:
        
        col_r2 = f"R2_{dataset}_{target}"
        col_mse = f"MSE_{dataset}_{target}"
        
        if col_r2 in results.columns:
            
            table = results[
                ["model", "Neurons", col_r2, col_mse]
            ].sort_values(by=col_r2, ascending=False)
            
            # Remove colchetes tratando como string
            for col in [col_r2, col_mse]:
                table[col] = (
                    table[col]
                    .astype(str)
                    .str.strip("[]")
                    .astype(float)
                )
            
            best_models_tables[f"{dataset}_{target}"] = table
            
            # 🔹 Calcula estatísticas
            mean_r2 = table[col_r2].mean()
            std_r2 = table[col_r2].std()
            mean_mse = table[col_mse].mean()
            std_mse = table[col_mse].std()
            
            # 🔹 Salva no resumo
            summary_rows.append({
                "dataset": dataset,
                "target": target,
                "mean_r2": mean_r2,
                "std_r2": std_r2,
                "mean_mse": mean_mse,
                "std_mse": std_mse
            })
            
            print(f"\n===== {dataset} - {target} =====")
            print(table.head(10))


# 🔹 Cria DataFrame final
df_summary = pd.DataFrame(summary_rows)

# 🔹 Exporta
df_summary.to_excel("Resumo_Estatisticas.xlsx", index=False)

print("\nResumo exportado com sucesso.")
```

```python
best_only = []
for dataset in SETS:
    for target in TARGETS:
        col_r2 = f"R2_{dataset}_{target}"
        
        if col_r2 in results.columns:
            
            idx_best = results[col_r2].idxmax()
            
            best_only.append({
                "Target": target,
                "Dataset": dataset,
                "Best_Model": results.loc[idx_best, "model"],
                "Neurons": results.loc[idx_best, "Neurons"],
                "Best_R2": results.loc[idx_best, col_r2]
            })

best_only_df = pd.DataFrame(best_only)

print("\n===== MELHOR MODELO POR TARGET/DATASET =====")
print(best_only_df)
```
